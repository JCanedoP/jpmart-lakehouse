```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS jpmart_ext_dl
    URL 'abfss://jpmart-container@jpmartextdl.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `jpmart-rg`)
```

```sql
CREATE CATALOG IF NOT EXISTS jpmart
        MANAGED LOCATION 'abfss://jpmart-container@jpmartextdl.dfs.core.windows.net/'
        COMMENT 'This is the catalog for the jpmart Data Lakehouse'
```

```sql
USE CATALOG jpmart;

CREATE SCHEMA IF NOT EXISTS bronze
        MANAGED LOCATION 'abfss://jpmart-container@jpmartextdl.dfs.core.windows.net/bronze'
```

```sql
USE CATALOG jpmart;
USE SCHEMA bronze;

CREATE EXTERNAL VOLUME IF NOT EXISTS raw_files
        LOCATION 'abfss://jpmart-container@jpmartextdl.dfs.core.windows.net/bronze/raw_files/'
```

```sql
USE CATALOG jpmart;

CREATE SCHEMA IF NOT EXISTS silver
    MANAGED LOCATION 'abfss://jpmart-container@jpmartextdl.dfs.core.windows.net/silver'
    COMMENT 'Cleaned and validated data';

CREATE SCHEMA IF NOT EXISTS gold
    MANAGED LOCATION 'abfss://jpmart-container@jpmartextdl.dfs.core.windows.net/gold'
    COMMENT 'Business-ready aggregated data';
```
